# High Resolution Snaps — Export Script

Copies all `G[R1]` files (full resolution pyramid level) from a folder
into a subfolder called `high resolution snaps`.

- **Copies** files — originals are not moved or touched
- Creates the subfolder automatically if it does not exist
- Skips the CZI, SlideLabel, and SlideOverview files

```python
import os
import shutil

# Set this to the folder you just renamed
folder_path = r"/Volumes/cnelya_lab/Lab/Analysis/Charlie_Heacock/E.Muscae Paraffin Slices - Slide Scanner/36"
```

```python
# ============================================================
# FIND ALL R1 FILES AND PREVIEW
# ============================================================

output_folder = os.path.join(folder_path, "High Resolution Snaps")

# Find all files containing G[R1] in the name
r1_files = sorted([
    f for f in os.listdir(folder_path)
    if os.path.isfile(os.path.join(folder_path, f))
    and "G[R1]" in f
    and not f.startswith("._")    # exclude Mac resource fork files # (macOS makes these hidden '._' copies on network drives)
])

if not r1_files:
    print("❌ No G[R1] files found — check that folder_path is correct "
          "and that files have been renamed already")
else:
    print(f"Found {len(r1_files)} G[R1] files\n")
    print(f"Destination folder : {output_folder}\n")
    print(f"{'Filename'}")
    print("-" * 80)
    for f in r1_files:
        size_mb = os.path.getsize(os.path.join(folder_path, f)) / (1024 ** 2)
        print(f"  {f:<70} {size_mb:>8.1f} MB")

    total_size = sum(
        os.path.getsize(os.path.join(folder_path, f)) for f in r1_files
    ) / (1024 ** 2)
    print(f"\n  Total size to copy : {total_size:.1f} MB")
    print("\n  → Run Cell 3 to copy these files")
```

```python
# ============================================================
# EXECUTE COPY
# ============================================================

# Create the output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

copied = 0
errors = []

for f in r1_files:
    src = os.path.join(folder_path, f)
    dst = os.path.join(output_folder, f)

    # Skip if a copy already exists there
    if os.path.exists(dst):
        print(f"⏭️  Already exists, skipped : {f}")
        continue

    try:
        shutil.copy2(src, dst)    # copy2 preserves file metadata/timestamps
        print(f"✅  Copied : {f}")
        copied += 1
    except Exception as e:
        print(f"❌  Failed : {f}  |  Error : {e}")
        errors.append((f, str(e)))

print(f"\n{'=' * 60}")
print(f"  Done — {copied} files copied, {len(errors)} errors")
print(f"  Saved to : {output_folder}")
```
